```python
valid_email_dup_mask = (
    (df["Email_Quality"] == "Valid") &
    (df["Email_Normalized"].notna()) &
    (df.groupby("Email_Normalized")["Email_Normalized"].transform("size") == 2)
)

valid_email_dups = df.loc[
    valid_email_dup_mask,
    ["Customer_ID", "First_Name", "Last_Name",
     "Email", "Email_Normalized", "Phone_Digits",
     "Date_of_Birth", "Country", "State",
     "Signup_Date", "Customer_Status"]
].sort_values("Email_Normalized")

valid_email_dups
```

```python
#Compare the 98 pairs of duplicated email records to see if they are the same customer.

match_cols = [
    "First_Name",
    "Last_Name",
    "Phone_Digits",
    "Date_of_Birth",
    "Country",
    "State",
    "Signup_Date",
    "Customer_Status"
]

duplicate_records = df.loc[
    df["Email_Normalized"].duplicated(keep=False)
    & df["Email_Normalized"].notna()
    & (df["Email_Quality"] == "Valid")
].copy()

# Compare each field within an email group
for col in match_cols:
    duplicate_records[f"{col}_Match"] = (
        duplicate_records.groupby("Email_Normalized")[col]
        .transform(lambda x: x.nunique(dropna=False) == 1)
    )

duplicate_records["All_Attributes_Match"] = duplicate_records[
    [f"{col}_Match" for col in match_cols]
].all(axis=1)

duplicate_records[
    ["Customer_ID", "Email_Normalized"] +
    [f"{col}_Match" for col in match_cols] +
    ["All_Attributes_Match"]
].head(20)
```

```python
pair_analysis = (
    duplicate_records
    .groupby("Email_Normalized")
    .agg(
        Records=("Customer_ID", "size"),
        Customer_IDs=("Customer_ID", lambda x: list(x)),
        First_Name=("First_Name", lambda x: x.nunique(dropna=False)),
        Last_Name=("Last_Name", lambda x: x.nunique(dropna=False)),
        Phone=("Phone_Digits", lambda x: x.nunique(dropna=False)),
        DOB=("Date_of_Birth", lambda x: x.nunique(dropna=False)),
        Country=("Country", lambda x: x.nunique(dropna=False)),
        State=("State", lambda x: x.nunique(dropna=False)),
        Signup_Date=("Signup_Date", lambda x: x.nunique(dropna=False)),
        Status=("Customer_Status", lambda x: x.nunique(dropna=False))
    )
    .reset_index()
)

pair_analysis.head(20)
```

```python
pair_analysis["Exact_Match"] = (
    (pair_analysis["Records"] == 2) &
    (pair_analysis["First_Name"] == 1) &
    (pair_analysis["Last_Name"] == 1) &
    (pair_analysis["Phone"] == 1) &
    (pair_analysis["DOB"] == 1) &
    (pair_analysis["Country"] == 1) &
    (pair_analysis["State"] == 1) &
    (pair_analysis["Signup_Date"] == 1) &
    (pair_analysis["Status"] == 1)
)

pair_analysis["Exact_Match"].value_counts()
```

So we have strong evidence that these are true duplicate customer records, not legitimate cases of two customers sharing an email.

```python
#Next step: create the duplicate mapping
#Which Customer_ID becomes the surviving/master ID?
#Retain the non-CUST010xxx Customer_ID as the master and mark the CUST010xxx record as the duplicate.

pair_analysis[["Customer_IDs"]].head(10)
```

```python
# Get all Customer IDs involved in duplicate pairs
duplicate_customer_ids = [
    customer_id
    for ids in pair_analysis["Customer_IDs"]
    for customer_id in ids
]

# Get the actual duplicate records
duplicate_records = df[
    df["Customer_ID"].isin(duplicate_customer_ids)
].copy()

# Keep one record from each duplicate pair
golden_records = (
    duplicate_records
    .drop_duplicates(subset=["Email_Normalized"])
    .copy()
)

# Create Master Customer IDs
golden_records["Master_Customer_ID"] = [
    f"MASTER{i:04d}"
    for i in range(1, len(golden_records) + 1)
]

golden_records.head()
```
